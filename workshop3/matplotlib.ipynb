# 2. Matplotlib Basics
---

## 2.1 Creating a Simple Plot

- To use Matplotlib we need to:
  - `pip install matplotlib`
  - `import matplotlib.pyplot as plt`

````python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)

plt.plot(x, np.sin(x))
plt.plot(x, np.cos(x))
````

---
## 2.2 Saving Figures to a File

- A Matplotlib `figure` can be saved to disk with the method `savefig()`.
- The file extension determines the file type (png, jpg, etc.).

````python
fig = plt.figure()

x = np.linspace(0, 10, 100)

plt.plot(x, np.sin(x))
plt.plot(x, np.cos(x))

fig.savefig('my_figure.png')
````

---
## 2.3 Matplotlib's Stateful Interface

- For simple plots, Matpotlib’s stateful interface can be used.
- A `plt.Figure` object is created with `plt.figure()`.
  - This becomes the **current** `figure`.
  - Plotting commands `plt.plot()` draw to the **current** `figure`.
- When `plt.figure()` is used again, a new `plt.Figure` object is created.
  - This becomes the new **current** `figure`.
  - Now all plotting commands draw to the new **current** `figure`.
- A reference to the **current** `figure` and `axes` can be obtained with:

    ```python
    plt.gcf() # get current figure
    plt.gca() # get current axes
    ```

---
## 2.4 Matplotlib's Object-Oriented Interface

- For general plots, Matpotlib’s object-oriented interface can be used.
- References to a `plt.Figure` and its `axes` are obtained.

  ```python
  fig, ax = plt.subplots(2) # creates two axes
  ```

- Then the `plt.Axes` object can be used for plotting commands.

  ```python
  ax[0].plot() # plot to the first axis
  ax[1].plot() # plot to the second axis
  ```

- The `fig` reference can also be used to configure the `plt.Figure` object.
- Greater control is obtained by using the object-oriented interface.

# 3. Simple Line Plots
---

## 3.1 Creating Simple Line Plots

- `plot(x,y)` plots a line with `x` and `y` values.
  - Using the object-oriented interface: `ax.plot(x,y)`.
  - Using the stateful interface: `plt.plot(x,y)`.
  
[https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.figure.html](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.figure.html)

[https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.axes.html](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.axes.html)

[https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.plot.html](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.plot.html)

````python
# object-oriented
fig = plt.figure()         # create figure
ax = plt.axes()            # create axes
x = np.linspace(0,10,1000) # create x values
ax.plot(x, np.sin(x))      # plot x=x, y=sin(x)
ax.plot(x, np.cos(x))      # plt x=x, y=cos(x)

# stateful
plt.figure()               # create figure
x = np.linspace(0,10,1000) # create x values
plt.plot(x, np.sin(x))     # plot x=x, y=sin(x)
plt.plot(x, np.cos(x))     # plot x=x, y=cos(x)
````

---
## 3.2 Adjusting the Plot: Line Colors

- `plot(x,y,color=)` accepts a `color` keyword argument.
- It accepts a name, code, grayscale, hex, tuple, HTML as a color.

[https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.plot.html](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.plot.html)

````python
plt.figure()

x = np.linspace(0,10,1000)

plt.plot(x, np.sin(x-0), color='blue')        # color by name
plt.plot(x, np.sin(x-1), color='g')           # color code (rgbcmyk)
plt.plot(x, np.sin(x-2), color='0.75')        # Grayscale (0-1.0)
plt.plot(x, np.sin(x-3), color='#FFDD44')   # Hex (RRGGBB, 00-FF)
plt.plot(x, np.sin(x-4), color=(1.0,0.2,0.3)) # RGB tuples (0-1.0)
plt.plot(x, np.sin(x-5), color='chartreuse')  # HTML color
````

---
## 3.3 Adjusting the Plot: Line Styles

- `plot(x,y,linestyle=)` accepts a `linestyle` argument.
- It accepts a line style name or a line style code.

[https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.plot.html](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.plot.html)

````python
plt.figure()

x = np.linspace(0,10,1000)

plt.plot(x, x + 0, linestyle='solid')   # solid line
plt.plot(x, x + 1, linestyle='dashed')  # dashed line
plt.plot(x, x + 2, linestyle='dashdot') # dashdot line
plt.plot(x, x + 3, linestyle='dotted'); # dotted line

# For short, you can use the following codes:
plt.plot(x, x + 4, linestyle='-')  # solid line
plt.plot(x, x + 5, linestyle='--') # dashed line
plt.plot(x, x + 6, linestyle='-.') # dashdot line
plt.plot(x, x + 7, linestyle=':'); # dotted line
````

---
## 3.4 Adjusting the Plot: Line Colors and Styles

- We can combine line color and linestyle codes.

[https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.plot.html](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.plot.html)

````python
plt.figure()

x = np.linspace(0,10,1000)

plt.plot(x, x + 0, '-g')  # solid green
plt.plot(x, x + 1, '--c') # dashed cyan
plt.plot(x, x + 2, '-.k') # dashdot black
plt.plot(x, x + 3, ':r'); # dotted red
````

---
## 3.5 Adjusting the Plot: Axes Limits

- `plt.xlim(start,stop)` and `plt.ylim(start,stop)` set axes limits (how much of the axis is shown).

[https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.xlim.html](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.xlim.html)

[https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.ylim.html](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.ylim.html)

````python
plt.figure()

x = np.linspace(0,10,1000)

plt.plot(x, np.sin(x))

plt.xlim(-1, 11)    # x between -1 and 11
plt.ylim(-1.5, 1.5) # y between -1.5 and 1.5
````

- `plt.axis([x_start,x_stop,y_start,y_stop])` sets both axes limits in one single command.

[https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.axes.html](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.axes.html)

````python
plt.figure()

x = np.linspace(0,10,1000)

plt.plot(x, np.sin(x))
plt.axis([-1, 11, -1.5, 1.5]) # x: -1 to 11, y: -1.5 to 1.5
````

## 3.6 Labeling Plots: Title and Axes

- `plt.title()`, `plt.xlabel()` and `plt.ylabel()` label the title, x axis and y axis, respectively.
- The position, size, and style can be set with optional arguments.

[https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.title.html](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.title.html)

[https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.xlabel.html](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.xlabel.html)

---

- `plt.plot()` has a `label` argument for labeling a plot.
- Labeled plots are displayed in a legend with `plt.legend()`.


[https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.plot.html](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.plot.html)

---
## 3.8 Stateful vs Object-Oriented Interface

- Most commands have the same name in the stateful and object-oriented interface.
- Some differ slightly, especially for setting limits, labels, and titles.

| Stateful       | Object-Oriented   |
|:--------------:|:-----------------:|
| `plt.xlabel()` | `ax.set_xlabel()` |
| `plt.ylabel()` | `ax.set_ylabel()` |
| `plt.xlim()`   | `ax.set_xlim()`   |
| `plt.ylim()`   | `ax.set_xlim()`   |
| `plt.title()`  | `ax.set_title()`  |

## 4.1 Simple Scatter Plots with `plt.plot`

- `plot(x,y,marker)` can also create scatter plots.
  - The third argument is a code representing the `marker` to print.

[https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.plot.html](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.plot.html)

````python
plt.figure()
x = np.linspace(0,10,30)
y = np.sin(x)
plt.plot(x, y, 'o', color='black') # circle marker (o)

plt.figure()
rng = np.random.RandomState(0)
markers = ['o','.',',','x','+','v','^','<','>','s','d']
for marker in markers:
    plt.plot(rng.rand(5), rng.rand(5), marker, # marker
             label="marker={0}".format(marker))

plt.legend(numpoints=1)
plt.xlim(0, 1.8)
````

- Marker codes can be combined with line and color codes.
- Additional arguments specify more properties for markers and lines.

````python
x = np.linspace(0,10,30)
y = np.sin(x)

plt.figure()
plt.plot(x, y, '-or') # line (-) circle marker (o) red (r)

plt.figure()
plt.plot(x, y, '-p', color='gray', linewidth=4,
         markersize=15, markerfacecolor='white',
         markeredgecolor='gray', markeredgewidth=2)
plt.ylim(-1.2, 1.2)
````

## 4.2 Simple Scatter Plots with `plt.scatter`

- `scatter(x,y,marker)` also creates scatter plots.
  - Uses the same marker codes as `plot(x,y,marker)`.
  - But properties for each individual point can be set with `scatter()`.
  - Keyword argument `alpha` sets the transparency (`0-1.0`), `0`=fully transparent.
  - Keyword argument `cmap` sets the color map (set of colors) to use.
  - `plt.colorbar()` displays a color bar next to the plot showing the color scale.
  
[https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.scatter.html](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.scatter.html)

````python
rng = np.random.RandomState(0)
x = rng.randn(100)
y = rng.randn(100)

plt.figure()
plt.scatter(x, y, marker='o') # circle marker (o)

plt.figure()
colors = rng.rand(100)
sizes = 1000 * rng.rand(100)
plt.scatter(x, y, c=colors, s=sizes, alpha=0.3, cmap='viridis')
plt.colorbar() # show color scale
````

## 5.1 Visualizing Discrete Errors

- `errorbar(x,y)` plots an error bar around each point.
  - Keyword arguments `xerr` and `yerr` represent the error.
  - Keyword argument `fmt` is equivalent to marker, line, and color codes in `plot(x,y)`.

[https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.errorbar.html](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.errorbar.html)

````python
plt.figure()
dx = 0.8 # x error varies between 0 to 0.8
x = np.linspace(0,10,50) + dx * np.random.randn(50)
y = np.sin(x)
plt.errorbar(x, y, xerr=dx, fmt='.k'); # horizontal error

plt.figure()
x = np.linspace(0,10,50)
dy = 0.8 # y error varies between 0 to 0.8
y = np.sin(x) + dy * np.random.randn(50)
plt.errorbar(x, y, yerr=dy, fmt='.k'); # vertical error
````

- Additional keyword arguments can be used to configure the errorbar.

````python
plt.figure()
x = np.linspace(0,10,50)
dy = 0.8
y = np.sin(x) + dy * np.random.randn(50)

plt.errorbar(x, y, yerr=dy,
fmt='o',            # marker type (circle o)
color='black',      # marker color
ecolor='lightgray', # error bar color
elinewidth=3,       # error bar line width
capsize=5);         # size of error bar ends (caps)
````

---
## 5.2 Visualizing Continuous Errors

- `fill_between(x,y_lower,y_upper)` is used for continuous errors.
  -  It fills the area between `y_lower` and `y_upper` with a color.

[https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.fill_between.html](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.fill_between.html)

````python
# create dataset
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import RBF, ConstantKernel as C

model = lambda x: x * np.sin(x)
xdata = np.array([1,3,5,6,8]) # x data for points
ydata = model(xdata)          # y data for points
kernel = C(1.0, (1e-3, 1e3)) * RBF(length_scale=1.0, length_scale_bounds=(1e-2, 1e2))
gp = GaussianProcessRegressor(kernel=kernel, n_restarts_optimizer=10)
gp.fit(xdata[:, np.newaxis], ydata)
x = np.linspace(0,10,1000)                                  # x data for line
y, sigma = gp.predict(x[:, np.newaxis], return_std=True) # y data for line
dy = 2 * sigma                                              # y error

# plot
plt.figure()
plt.plot(xdata, ydata, 'or')      # plot points
plt.plot(x, y, '-', color='gray') # plot line
plt.fill_between(x, y-dy, y+dy, color='gray', alpha=0.2) # fill area between y – dy (lower)
                                                         # and y + dy (upper)
plt.xlim(0, 10);
````

# 6. Density and Contour Plots

## 6.1 Creating Contour Plots with `plt.contour`

- `contour(X,Y,Z)` plots a contour from 2D arrays `X`, `Y` and `Z`.
  - It’s used to visualize 3D data in 2D (2 dimensions).
  - `X,Y=np.meashgrid(x,y)` creates a grid of 2D arrays `X` and `Y` from 1D arrays `x` and `y`, where `Z=f(X,Y)` creates a 2D array `Z`.
  - The 2D arrays can then be used in `contour(X,Y,Z)`.
- In a contour plot:
  - solid lines are used for positive values.
  - dotted lines are used for negative values. 

[https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.contour.html](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.contour.html)

[https://numpy.org/doc/2.2/reference/generated/numpy.meshgrid.html](https://numpy.org/doc/2.2/reference/generated/numpy.meshgrid.html)

## 7.1 Plotting Histograms with `plt.hist`

- `hist(data)` plots a histogram from 1D `data`.
- Additional keyword arguments configure its appearance, e.g.
  - `bins=10` (number of bins)
  - `density=True` (normalize area to `1.0`)
  - `histtype='stepfilled'` (type of histogram, here filled steps)

[https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.hist.html](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.hist.html)

````python
data = np.random.randn(1000)

plt.figure()
plt.hist(data) # plot histogram

plt.figure()
plt.hist(data, bins=30, density=True, # additional settings
         alpha=0.5, histtype='stepfilled',
         color='steelblue', edgecolor='none')
````

- `histtype='stepfilled'` with `alpha=0.3` is good for comparing multiple histograms in one figure.

````python
x1 = np.random.normal(0, 0.8, 1000)
x2 = np.random.normal(-2, 1, 1000)
x3 = np.random.normal(3, 2, 1000)

plt.figure()

# use these keyword arguments for all three plots
kwargs = dict(histtype='stepfilled', alpha=0.3, density=True, bins=40)

plt.hist(x1, **kwargs) # first histogram plot
plt.hist(x2, **kwargs) # second histogram plot
plt.hist(x3, **kwargs) # third histogram plot
````

## 9.1 Creating a Colorbar

- `colorbar()` creates a bar with a labeled scale of colors.
  - It’s used as key for the meaning of colors in a plot.
  - By default, it’s placed to the right of the main plot.
  - `imshow()` interprets an array of numbers as an image.
  - It uses the default colormap if no colors are specified.
  - The default colormap is `'viridis'` (a set of blue and green colors).

[https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.colorbar.html](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.colorbar.html)

[https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.imshow.html](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.imshow.html)

````python
# create dataset
x = np.linspace(0, 10, 1000)
I = np.sin(x) * np.cos(x[:, np.newaxis])

plt.figure()
plt.imshow(I)  # plot image using default colormap
plt.colorbar() # plot colorbar
````

---
## 9.2 Colormaps

- Keyword `cmap` specifies the colormap to use in a plot.
- There are three different categories of colormaps:
  - **Sequential**, e.g. `cmap='viridis'`, a continuous sequence of colors.
  - **Divergent**, e.g. `cmap='RdBu'`, usually contain two distinct colors.
  - **Qualitative**, e.g. `cmap='jet'`, a non-sequential mix of colors.

[https://matplotlib.org/stable/gallery/color/colormap_reference.html](https://matplotlib.org/stable/gallery/color/colormap_reference.html)

````python
x = np.linspace(0, 10, 1000)
I = np.sin(x) * np.cos(x[:, np.newaxis])

plt.figure()
plt.imshow(I, cmap='viridis') # sequence of blue to green
plt.colorbar()                # plot colorbar

plt.figure()
plt.imshow(I, cmap='RdBu')    # distinct colors red (Rd) and blue (Bu)
plt.colorbar()                # plot colorbar

plt.figure()
plt.imshow(I, cmap='jet')     # non-sequential mix of colors
plt.colorbar()                # plot colorbar
````

---
## 9.3 Sampling from a Colormap

- `plt.get_cmap('Blues', 6)` samples a number of colors from a colormap.
  - Most colormaps are a continuous sequence of colors.
  - Sampling a set of colors means we can create a discrete set of colors.
  - `plt.clim(lower,upper)` sets a `lower` and `upper` limit for a colorbar.

[https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.get_cmap.html](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.get_cmap.html)

[https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.clim.html](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.clim.html)

````python
x = np.linspace(0, 10, 1000)
I = np.sin(x) * np.cos(x[:, np.newaxis])

plt.figure()
plt.imshow(I, cmap=plt.get_cmap('Blues', 6)) # sample 6 colors
plt.colorbar()

plt.figure()
plt.imshow(I, cmap=plt.get_cmap('Blues', 6)) # sample 6 colors
plt.colorbar()
plt.clim(-1, 1) # set colorbar limits: -1 to +1
````

# 10. Multiple Subplots

## 10.1 Creating Subplots with `plt.subplot`

- A subplot is a `plt.Axes` within a `plt.Figure`.
- Subplots can be added to a figure using the stateful or object-oriented interface.
  - Stateful: `plt.axes([bottom, left, width, height])`
  - Object-oriented: `fig.add_axes([bottom, left, width, height])`
- `bottom`, `left`, `width`, and `height` are values in the normalized figure coordinate system.
  - The normalized figure has `bottom=0`, `left=0`, `width=1`, and `height=1`.

[https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.axes.html](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.axes.html)

[https://matplotlib.org/stable/api/_as_gen/matplotlib.figure.Figure.add_axes.html](https://matplotlib.org/stable/api/_as_gen/matplotlib.figure.Figure.add_axes.html)

````python
x = np.linspace(0, 10)

# Stateful Interface
plt.figure()
ax1 = plt.axes([0.0, 0.0, 1.0, 1.0])   # axis1 covering entire figure
ax2 = plt.axes([0.65, 0.65, 0.2, 0.2]) # axis2 at (0.65,0.65) width=height=0.2
ax1.plot(np.sin(x))
ax2.plot(np.cos(x))

# Object-Oriented Interface
fig = plt.figure()
ax1 = fig.add_axes([0.0, 0.0, 1.0, 1.0])   # axis1
ax2 = fig.add_axes([0.65, 0.65, 0.2, 0.2]) # axis2
ax1.plot(np.sin(x))
ax2.plot(np.cos(x))
````

- `plt.subplot(rows,cols,index)` creates a subplot
within a grid of subplots with aligned rows and columns.
  - `rows` is the number of rows in the grid.
  - `cols` is the number of columns in the grid.
  - `index` is the subplot’s index within the grid.

[https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.subplot.html](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.subplot.html)

````python
# Stateful interface
plt.figure()

for i in range(1, 7):
    plt.subplot(2, 3, i) # create subplot
    plt.text(0.5, 0.5, str((2, 3, i)), fontsize=18, ha='center') # text
````

---
## 10.2 Adjusting Subplots

- `plt.subplots_adjust()` adjusts the subplots in a grid.
  - `hspace=0.4` adds space between each row (h=height).
  - `wspace=0.4` adds space between each column (w=width).

[https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.subplot.html](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.subplot.html)

[https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.subplots_adjust.html](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.subplots_adjust.html)

````python
# Stateful interface
plt.figure()
plt.subplots_adjust(hspace=0.4, wspace=0.4) # height and and width spacing

for i in range(1, 7):
    plt.subplot(2, 3, i) # create subplot
    plt.text(0.5, 0.5, str((2, 3, i)), fontsize=18, ha='center') # text
````

---
## 10.3 Using the Object-Oriented Interface

- The object-oriented interface:
  - Creates a `plt.Figure` object with a reference in variable `fig=figure()`.
  - Calls `fig.add_subplot()` to add a `plt.Axes` object to the grid.
  - Calls `fig.subplots_adjust()` to adjust the subplots in the figure.

[https://matplotlib.org/stable/api/_as_gen/matplotlib.figure.Figure.add_subplot.html](https://matplotlib.org/stable/api/_as_gen/matplotlib.figure.Figure.add_subplot.html)

[https://matplotlib.org/stable/api/_as_gen/matplotlib.figure.Figure.subplots_adjust.html](https://matplotlib.org/stable/api/_as_gen/matplotlib.figure.Figure.subplots_adjust.html)

````python
# Object-oriented interface
fig = plt.figure()
fig.subplots_adjust(hspace=0.4, wspace=0.4) # height and width spacing

for i in range(1, 7):
    ax = fig.add_subplot(2, 3, i) # create subplot
    ax.text(0.5, 0.5, str((2, 3, i)), fontsize=18, ha='center') # text
````

## 10.4 Creating Subplots with `plt.subplots`

- `fig, ax = plt.subplots(rows,cols)` creates a whole
`rows x cols` grid at once, returning the figure `fig` and axes `ax`.
  - `fig` is a reference to a `plt.Figure` object.
  - `ax` is a reference to a `plt.Axes` object if `rows=cols=1`.
    - else a `[rows x cols]` array of `plt.Axes` objects.

[https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.subplots.html](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.subplots.html)

````python
# create 2x3 grid
fig, ax = plt.subplots(2, 3)

# axes are in a two-dimensional array, indexed by ax[row, col]
for i in range(2):
    for j in range(3):
        ax[i, j].text(0.5, 0.5, str((i, j)), fontsize=18, ha='center') # text
````

- `fig, ax = plt.subplots(rows,cols)` has keyword arguments:
  - `sharex='col'` which let the subplots in a column share one x axis.
  - `sharey='row'` which let the subplots in a row share one y axis.
  - This makes the plot cleaner, since each axis isn’t shown for every subplot.

````python
# create 2x3 grid
fig, ax = plt.subplots(2, 3, sharex='col', sharey='row')

# axes are in a two-dimensional array, indexed by ax[row, col]
for i in range(2):
    for j in range(3):
        ax[i, j].text(0.5, 0.5, str((i, j)), fontsize=18, ha='center') # text
````

## 10.5 Creating Subplots with `plt.GridSpec`

- `grid = plt.GridSpec(rows,cols)` can be used for more complicated grid layouts
- Indexing with ranges in the returned `grid[]` array permits creating subplots spanning over multiple rows and/or columns.

[https://matplotlib.org/stable/api/_as_gen/matplotlib.gridspec.GridSpec.html](https://matplotlib.org/stable/api/_as_gen/matplotlib.gridspec.GridSpec.html)

````python
# create 2x4 grid (with spacing)
grid = plt.GridSpec(2, 4, wspace=0.4, hspace=0.3)

ax1 = plt.subplot(grid[:2, 0])  # both rows, first column (0-1,0)
ax2 = plt.subplot(grid[0, 1])   # first row, second column (0,1)
ax3 = plt.subplot(grid[0, 2:])  # first row, third and fourth column (0,2-3)
ax4 = plt.subplot(grid[1, 1:3]) # second row, second and third column (1,1-2)
ax5 = plt.subplot(grid[1, 3]);  # second row, fourth column (1,3)

ax1.text(0.5, 0.5, '(0-1,0)', fontsize=18, ha='center') # draw text (0-1,0)
ax2.text(0.5, 0.5, '(0,1)', fontsize=18, ha='center')   # draw text (0,1)
ax3.text(0.5, 0.5, '(0,2-3)', fontsize=18, ha='center') # draw text (0,2-3)
ax4.text(0.5, 0.5, '(1,1-2)', fontsize=18, ha='center') # draw text (1,1-2)
ax5.text(0.5, 0.5, '(1,3)', fontsize=18, ha='center')   # draw text (1,3)
````

# 11. Text and Annotation

## 11.1 Annotations and Coordinate Systems

- `plt.text(x,y)` or `ax.text(x,y)` draws text at coordinate `(x,y)`.
  - By default, the text if drawn to the right of `(x,y)`.
- The coordinate system to use for `(x,y)` can be specified.
  - `transform=ax.transData` uses the data’s coord. system (default).
  - `transform=ax.transAxes` uses the axis’ coord. system.
  - `transform=fig.transFigure` uses the figure’s coord. system.
- A normalized coordinate system is used for the axis and figure (`0-1.0`).
- In the plot below, the origin `(0,0)` is located:
  - In the bottom left of the gray area for the `figure`.
  - In the bottom left of the white area for the `axis` and the `data`.

[https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.text.html](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.text.html)

[https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.text.html](https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.text.html)

````python
fig, ax = plt.subplots(facecolor='lightgray')
ax.axis([0, 10, 0, 10])

# transform=ax.transData is the default, but we'll specify it anyway
ax.text(1, 5, '. Data: (1, 5)', transform=ax.transData)              # data coord
ax.text(0.5, 0.1, '. Axes: (0.5, 0.1)', transform=ax.transAxes)      # ax coord
ax.text(0.2, 0.2, '. Figure: (0.2, 0.2)', transform=fig.transFigure) # fig coord
````

- Changing the axis limits `ax.set_xlim()`, `ax.set_ylim()`
  - Causes the location of a point in the data’s coord. system to move.
  - Doesn’t affect the location of a point in the axis’ coord. system.
  - Doesn’t affect the location of a point in the figure’s coord. System.

````python
fig, ax = plt.subplots(facecolor='lightgray')
ax.axis([0, 10, 0, 10])

# transform=ax.transData is the default, but we'll specify it anyway
ax.text(1, 5, '. Data: (1, 5)', transform=ax.transData)              # data coord
ax.text(0.5, 0.1, '. Axes: (0.5, 0.1)', transform=ax.transAxes)      # ax coord
ax.text(0.2, 0.2, '. Figure: (0.2, 0.2)', transform=fig.transFigure) # fig coord
ax.set_xlim(0, 2)  # set x axis: 0 to 2
ax.set_ylim(-6, 6) # set y axis: -6 to 6
````

## 11.2 Text

- `ax.text(x,y,text)` draws text using the data’s coord. (default).
  - Keyword argument `size` sets the text size, e.g. `size=10`.
  - Keyword argument `color` sets the text color, e.g. `color='gray'`.
  - Keyword argument `ha` sets the horizontal alignment w.r.t. `(x,y)`.
    - `ha='center'` → text centered around `(x,y)`
    - `ha='left'`   → `(x,y)` left of text
    - `ha='right'`  → `(x,y)` right of text
  - There is also a `va='center'`, `va='bottom'`, `va='top'`.

[https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.text.html](https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.text.html)

````python
fig, ax = plt.subplots()
x = np.linspace(0, 20, 1000)

ax.plot(x, np.cos(x))
ax.axis('equal') # make scale (aspect ratio) on both axes equal

ax.text(0, 2, 'none', size=10, color='gray')                 # no horizontal alignment
ax.text(3, -2, 'center', size=10, color='gray', ha='center') # center text on (3, -2)
ax.text(12.5, 2, 'left', size=10, color='gray', ha='left')   # (12.5, 2) left of text
ax.text(18.8, 2, 'right', size=10, color='gray', ha='right') # (18.8, 2) right of text
````

## 11.2 Annotate

- `ax.annotate(text,xy=(x,y))` also draws text using the data’s coord.
  - It (almost) has the same keyword arguments as `text(x,y,text)`
    - But instead uses `xy=(x,y)` to set the coordinates.

[https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.annotate.html](https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.annotate.html)

````python
fig, ax = plt.subplots()
x = np.linspace(0, 20, 1000)

ax.plot(x, np.cos(x))
ax.axis('equal') # make scale (aspect ratio) on both axes equal

ax.annotate('none', xy=(0, 2), size=10, color='gray')
ax.annotate('center', xy=(3, -2), size=10, color='gray', ha='center')
ax.annotate('left', xy=(12.5, 2), size=10, color='gray', ha='left')
ax.annotate('right', xy=(18.8, 2), size=10, color='gray', ha='right')
````

- `ax.annotate(text,xy=(x,y))` accepts additional keyword arguments.
  - `xy=(x,y)` is the same as before.
  - `xycoords='data'` uses the data’s coord. system for `xy` (default).
  - `xytext=(xt,yt)` draws the text at `(xt,yt)` instead.
  - `textcoords='data'` uses the data’s coord. system for `xytext`.
    - `textcoords='offset points'` uses the text’s coord. system.

````python
fig, ax = plt.subplots()
x = np.linspace(0, 20, 1000)
ax.plot(x, np.cos(x))
ax.axis('equal')

ax.annotate('none', size=10, color='gray',
            xy=(0, 2))

ax.annotate('none', size=10, color='red',   # default for xycoords and textcoords
            xy=(0, 2),                      # the data point’s coordinates
            xytext=(0, 6))                  # the text’s coordinates

ax.annotate('none', size=10, color='green', # using textcoords='offset points'
            xy=(0, 2),
            xytext=(0, 50), textcoords='offset points') # text offset by (0, 50)
````

- To complete the example, we also need to label the ticks.
- `plt.FuncFormatter(func)` uses the provided function `func` to label ticks.
  - `func(x,pos)` takes a tick value `x` and a position `pos`, returning a formatted label.

[https://matplotlib.org/stable/api/ticker_api.html#matplotlib.ticker.FuncFormatter](https://matplotlib.org/stable/api/ticker_api.html#matplotlib.ticker.FuncFormatter)


````python
# Plot a sine and cosine curve
fig, ax = plt.subplots()
x = np.linspace(0, 3 * np.pi, 1000)
ax.plot(x, np.sin(x), lw=3, label='Sine')
ax.plot(x, np.cos(x), lw=3, label='Cosine')

# Set up grid, legend, and limits
ax.grid(True)
ax.legend(frameon=False)
ax.axis('equal')
ax.set_xlim(0, 3 * np.pi)

# Set x axis major and minor locator
ax.xaxis.set_major_locator(plt.MultipleLocator(np.pi / 2)) # display major ticks at multiples of pi/2
ax.xaxis.set_minor_locator(plt.MultipleLocator(np.pi / 4)) # display minor ticks at multiples of pi/4

# Define a function for FuncFormatter (it uses Matplotlib’s built-in support for LaTeX in strings)
def format_func(value, tick_number):
    # find number of multiples of pi/2
    N = int(np.round(2 * value / np.pi))
    if N == 0: return "0"
    elif N == 1: return r"$\pi/2$"
    elif N == 2: return r"$\pi$"
    elif N % 2 > 0: return r"${0}\pi/2$".format(N)
    else: return r"${0}\pi$".format(N // 2)

# Set x axis major tick formatter
ax.xaxis.set_major_formatter(plt.FuncFormatter(format_func)) # use FuncFormatter with format_func to format the labels
````

## 14.3 Three-Dimensional Contour Plots

- `contour3D(X,Y,Z)` is the equivalent of `contour(X,Y)`, but for 3D plots.
- `view_init(elevation,azimuth)` can be used to change the viewing angle for any 3D plot.
  - `elevation` is the number of degrees above the x-y plane (default: `30`).
  - `azimuth` is the counter-clockwise rotation in degrees about the z axis (default: `-60`).

[https://matplotlib.org/stable/gallery/mplot3d/index.html](https://matplotlib.org/stable/gallery/mplot3d/index.html)

````python
def f(x, y):
    return np.sin(np.sqrt(x ** 2 + y ** 2))

x = np.linspace(-6, 6, 30)
y = np.linspace(-6, 6, 30)
X, Y = np.meshgrid(x, y)
Z = f(X, Y)

fig = plt.figure()
ax = plt.axes(projection='3d')           # 3D axes
ax.contour3D(X, Y, Z, 50, cmap='binary') # 3D contour plot
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')

fig = plt.figure()
ax = plt.axes(projection='3d')           # 3D axes
ax.contour3D(X, Y, Z, 50, cmap='binary') # 3D contour plot
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
ax.view_init(60, 35)                     # view angle: elevation 60, azimuth 35
````

````python
def f(x, y):
    return np.sin(np.sqrt(x ** 2 + y ** 2))

x = np.linspace(-6, 6, 30)
y = np.linspace(-6, 6, 30)
X, Y = np.meshgrid(x, y)
Z = f(X, Y)

# Plot wrire frame
fig = plt.figure()
ax = plt.axes(projection='3d') # 3D axes
ax.plot_wireframe(X, Y, Z, color='black')
ax.set_title('wireframe')

# Plot surface
fig = plt.figure()
ax = plt.axes(projection='3d') # 3D axes
ax.plot_surface(X, Y, Z, rstride=1, cstride=1, cmap='viridis', edgecolor='none')
ax.set_title('surface')
````

## 15.2 Cylindrical Projections

- `projection='cyl'` uses a Cylindrical projection, with appropriate additional keyword arguments.
- A help function `draw_map()` is used to overlay a topography with lat/long lines.

[https://matplotlib.org/basemap/stable](https://matplotlib.org/basemap/stable)

[https://matplotlib.org/basemap/stable/users/examples.html](https://matplotlib.org/basemap/stable/users/examples.html)

````python
# Import chain funtion from itertools package
from itertools import chain

# Define helper function to overlay topography and draw lat/long lines
def draw_map(m, scale=0.2):
    # draw a shaded-relief image
    m.shadedrelief(scale=scale)
    
    # lats and longs are returned as a dictionary
    lats = m.drawparallels(np.linspace(-90, 90, 13))
    lons = m.drawmeridians(np.linspace(-180, 180, 13))
    
    # keys contain the plt.Line2D instances
    lat_lines = chain(*(tup[1][0] for tup in lats.items()))
    lon_lines = chain(*(tup[1][0] for tup in lons.items()))
    all_lines = chain(lat_lines, lon_lines)
    
    # cycle through these lines and set the desired style
    for line in all_lines:
        line.set(linestyle='-', alpha=0.3, color='w')

# Create figure (width: 8 inches, height: 6 inches)
fig = plt.figure(figsize=(8, 6), edgecolor='w')

# Create Basemap instance with Cylindrical projection
m = Basemap(projection='cyl', resolution=None, llcrnrlat=-90, urcrnrlat=90, llcrnrlon=-180, urcrnrlon=180)

# Call helper function
draw_map(m)
````

## 16.2 Histograms, KDEs, and Densities

- We have seen Matplotlib’s `plt.hist(data)`, used to create histogram plots.
- Seaborn’s `sns.kdeplot(data)` creates Kernel Density Estimation (KDE) plots.
  - Keyword argument `fill=True` fills the area under the distribution curves.

[https://seaborn.pydata.org](https://seaborn.pydata.org)

````python
# Create some data
import pandas as pd
np.random.seed(0)
data = np.random.multivariate_normal([0,0],[[5,2],[2,2]], size=2000)
data = pd.DataFrame(data, columns=['x', 'y'])

# Plot histogram using Matplotlib’s plt.hist
plt.figure()
for col in 'xy':
    plt.hist(data[col], density=True, alpha=0.5)

# Plot KDE using Seaborn’s sns.kdeplot
plt.figure()
for col in 'xy':
    sns.kdeplot(data[col], fill=True)
````

- Histograms and KDEs can be combined using Seaborn’s `sns.histplot(data)`.
- If we pass both `x` and `y` data to `sns.kdeplot(data)`, we get a contour plot.

````python
# Create some data
np.random.seed(0)
data = np.random.multivariate_normal([0,0],[[5,2],[2,2]], size=2000)
data = pd.DataFrame(data, columns=['x', 'y'])

# Plot histogram using Seaborn’s sns.histplot
plt.figure()
sns.histplot(data['x'], kde=True, stat='density', alpha=0.5)
sns.histplot(data['y'], kde=True, stat='density', alpha=0.5)

# Plot contour using Seaborn’s sns.kdeplot
plt.figure()
sns.kdeplot(x=data['x'], y=data['y'])
````

- We can see joint and marginal distributions together using Seaborn’s `sns.jointplot()`.
- Either as a KDE with `kind='kde'` or with hexagonal tiles with `kind='hex'`.

````python
# Create some data
np.random.seed(0)
data = np.random.multivariate_normal([0,0],[[5,2],[2,2]], size=2000)
data = pd.DataFrame(data, columns=['x', 'y'])

# Plot joint and marginal distributions as KDE
plt.figure()
with sns.axes_style('white'):
    sns.jointplot(x='x', y='y', data=data, kind='kde',
                  joint_kws=dict(fill=True))

# Plot joint and marginal distributions as HEX
plt.figure()
with sns.axes_style('white'):
    sns.jointplot(x='x', y='y', data=data, kind='hex')
````

## 16.4 Faceted Histograms

- Sometimes it’s better to view data as histograms of subsets.
- Seaborn’s `grid=sns.FacetGrid(data)` creates a grid of subsets.
  - `grid.map()` maps a function, here `plt.hist`, to the subsets.
  - Here we are using a Pandas DataFrame as the `data`.

[https://seaborn.pydata.org](https://seaborn.pydata.org)

````python
# Load Tips dataset, add a column, and print first 5 rows
tips = sns.load_dataset('tips')
tips['tip_pct'] = 100 * tips['tip'] / tips['total_bill']
print(tips.head())

# Create a sns.FacetGrid and map plt.hist to its subsets
plt.figure()
grid = sns.FacetGrid(tips, row='sex', col='time', margin_titles=True)
grid.map(plt.hist, 'tip_pct', bins=np.linspace(0, 40, 15));
````
